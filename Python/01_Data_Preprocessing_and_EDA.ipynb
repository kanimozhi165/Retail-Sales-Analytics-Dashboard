# 5. Exploratory Data Analysis (EDA)

## 5.0 Overall Business Metrics

```python
total_sales = df['sales'].sum()
total_profit = df['profit'].sum()
total_quantity = df['quantity'].sum()
print(f"Total Sales : {total_sales:,.2f}")
print(f"Total Profit : {total_profit:,.2f}")
print(f"Total Quantity : {total_quantity:,.0f}")
```

### Business Insight

The dataset contains a substantial volume of sales, profit, and quantity transactions, providing a strong foundation for analyzing business performance and identifying trends.

## 5.1 Sales by Category

**Business Question**

Which category generated the highest sales?

```python
category_sales = df.groupby('category')['sales'].sum()
category_sales.round(2)
```

### Business Insight

Technology generated the highest total sales, while Office Supplies generated the lowest.

This indicates that Technology products contribute the most to the company's overall revenue.

## 5.2 Profit by Category

### Business Question

Which product category generated the highest total profit?

```python
category_profit = df.groupby('category')['profit'].sum()
category_profit.round(2)
```

### Business Insight

Technology generated the highest total profit, while Furniture generated the lowest total profit.

This suggests that Furniture may have lower profit margins or higher operational costs compared to other categories.

## 5.3 Customer Analysis

### Business Question

Which customers generated the highest total sales?

```python
customer_sales = df.groupby('customer_name')['sales'].sum()
top_10_customers = customer_sales.sort_values(ascending=False).head(10)
top_10_customers
```

### Business Insight

Tom Ashbrook generated the highest total sales, making them one of the company's most valuable customers.

Retaining high-value customers is important for sustaining long-term revenue.

### Business Question

Which customers generated the highest total profit?

```python
customer_profit = df.groupby('customer_name')['profit'].sum()
top_10_customer_profit = customer_profit.sort_values(ascending=False).head(10)
top_10_customer_profit.round(2)
```

### Business Insight

Tamara Chand generated the highest total profit, indicating that not all high-sales customers are necessarily the most profitable.

## 5.4 Product Analysis

### Business Question

Which products generated the highest total sales?

```python
product_sales = df.groupby('product_name')['sales'].sum()
top_10_products = product_sales.sort_values(ascending=False).head(10)
top_10_products
```

### Business Insight

Apple Smart Phone, Full Size generated the highest total sales, making it the company's top revenue-generating product.

### Business Question

Which products generated the highest total profit?

```python
# "Is the highest-selling product necessarily the most profitable product?"
product_profit = df.groupby('product_name')['profit'].sum()
top_10_profit_product = product_profit.sort_values(ascending=False).head(10)
top_10_profit_product
```

### Business Insight

Canon imageCLASS 2200 Advanced Copier generated the highest total profit.

This shows that the highest-selling product is not always the most profitable product.

## 5.5 Regional Analysis

### Business Question

Which region generated the highest total sales?

```python
region_sales = df.groupby('region')['sales'].sum()
top_region_sales = region_sales.sort_values(ascending=False)
top_region_sales
```

```python
# Number of Unique values.
df["region"].nunique()
```

### Business Insight

The Central region generated the highest total sales, while Canada recorded the lowest sales.

This suggests stronger market performance in the Central region compared to other regions.

## 5.6 Average Sales Analysis

### Business Question

Which category has the highest average sales per order?

```python
category_avg_sales = df.groupby('category')['sales'].mean()
highest_avg_sales = category_avg_sales.sort_values(ascending=False)
highest_avg_sales
```

### Business Insight

Technology recorded the highest average sales per order, indicating that customers purchasing Technology products tend to place higher-value orders.

## 5.7 Order Analysis

### Business Question

Which category has the highest number of orders?

```python
category_order_count = df.groupby('category')['sales'].count()
category_order_count = category_order_count.sort_values(ascending=False)
category_order_count
```

### Business Insight

Office Supplies had the highest number of orders, showing strong customer demand even though its total sales were lower than Technology.

A higher order count does not necessarily result in the highest revenue.

## 5.8 Filtering

### Business Question

Show all orders where Sales is greater than 1000.

```python
highest_sales = df[df['sales']>1000]
highest_sales
```

### Business Insight

A total of 2,622 orders had sales greater than 1,000.

These high-value transactions contribute significantly to overall revenue and may represent premium products or bulk purchases.

### Business Question
Show all loss-making orders and determine what percentage of the total dataset they represent.

```python
lowest_profit = df[df['profit']<0]
lowest_profit
```

```python
# Calculate the percentage of loss-making orders
total_rows = df.shape[0]
loss_rows = lowest_profit.shape[0]
round((loss_rows/total_rows)*100,2)
```

### Business Insight

Approximately 24.46% of all orders resulted in a loss.

This indicates that nearly one-quarter of the transactions were unprofitable and should be investigated for pricing, discounts, or operational costs.

### Business Question

Show all Technology orders where Sales is greater than 1000.

```python
high_tech_products = df[(df['category']=='Technology') & (df['sales']>1000)]
high_tech_products
```

### Business Insight

Many high-value orders belong to the Technology category, reinforcing its position as the company's strongest revenue-generating category.

### Business Question

Show all orders where the category is either Technology or Furniture.

```python
tech_fur_details = df[(df['category']=='Technology') | (df['category']=='Furniture')]
tech_fur_details
```

### Business Insight

Technology and Furniture together account for a significant portion of the dataset, making them important categories for further sales analysis.

### Business Question

Show all orders from the Central region where Profit is greater than 500.

```python
high_central_profit = df[(df['region']=='central') & (df['profit']>500)]
high_central_profit
```

### Business Insight

The Central region contains several highly profitable orders, indicating strong business performance in this region.

### Business Question

Show all Technology orders from the Central region where Profit is greater than 500.

```python
high_tech_central_profit = df[(df['category']=='Technology') & (df['region']=='Central') & (df['profit']>500)]
high_tech_central_profit[["category", "region", "profit"]]
```

### Business Insight

Technology products sold in the Central region include many highly profitable orders, highlighting a combination of strong product performance and regional demand.

## 5.9 Time-Based Analysis

### Business Question

Which year generated the highest total sales?

```python
highest_sales_year = df.groupby('year')['sales'].sum()
highest_sales_year.sort_values(ascending=False)
```

### Business Insight

Sales increased over the years, with 2014 generating the highest revenue.

This indicates consistent business growth during the analyzed period.

### Business Question

Which month generated the highest total sales?

```python
df["month_name"] = df["order_date"].dt.month_name()
high_sales_month = df.groupby('month_name')['sales'].sum().sort_values(ascending=False)
high_sales_month
```

### Business Insight

December generated the highest total sales, suggesting strong seasonal demand during the year-end period.

February recorded the lowest sales among all months.

## 5.10 Multi-Level Grouping

### Business Question

Which category performs best in each region?

```python
best_category_region = df.groupby(['category', 'region'])['sales'].sum().sort_values(ascending=False)
best_category_region.reset_index()
```

### Business Insight

Technology generated the highest sales across most regions, demonstrating consistently strong performance regardless of location.

## 5.11 Pivot Table Analysis

### Business Question

Create a pivot table showing total sales for each category across all regions.

```python
pivot_sales = pd.pivot_table (
    df,
    index = 'category',
    columns = 'region',
    values = 'sales',
    aggfunc = 'sum'
)
pivot_sales
```

```python
pivot_sales["Total"] = pivot_sales.sum(axis=1)
pivot_sales = pivot_sales.sort_values(by="Total", ascending=False)
pivot_sales
```

### Business Insight

The pivot table provides a clear comparison of category performance across different regions, making it easier to identify the strongest product category within each region.

# 6. Data Visualization

```python
import matplotlib.pyplot as plt
```

## 6.1 Bar Chart

### Business Question

How do total sales compare across product categories?

```python
plt.figure(figsize=(8,5))

bars = plt.bar(category_sales.index, category_sales.values, color="green")

for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        f"{bar.get_height():,.0f}",
        ha="center",
        va="bottom"
    )

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.xticks(rotation=45) 

plt.grid(axis="y") 

plt.show()
```

## 6.2 Line Chart

### Business Question

How have total sales changed over the years?

```python
plt.figure(figsize=(8,5))

plt.plot(highest_sales_year.index, highest_sales_year.values, marker="o",linestyle="-.", color="red") 

plt.title('Total Sales by Year')

plt.xlabel('Year')
plt.ylabel('Total Sales')

plt.grid(axis='y')

plt.show()
```

## 6.3 Pie Chart

### Business Question

What percentage of total sales comes from each category?

```python
explode = (0, 0, 0.1)
plt.figure(figsize=(7,7))

plt.pie(
    category_sales.values,
    labels=category_sales.index,
    autopct="%1.1f%%",
    startangle=90,
    explode = explode,
    shadow=True
)
plt.title("Sales Distribution by Category")

plt.show()
```

## 6.4 Histogram

### Business Question

How are sales amounts distributed?

```python
plt.figure(figsize=(8,5))

plt.hist(df["sales"],bins=40, edgecolor="black")

plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency") 

plt.show()
```

### Observation

The sales distribution is right-skewed, indicating that most orders have relatively low sales while a small number of orders have very high sales.

## 6.5 Scatter Plot

### Business Question

Is there a relationship between sales and profit?

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["sales"],
    df["profit"],
    alpha=0.5, 
    s=10 
)
plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.show()
```

### Observation

The scatter plot shows a generally positive relationship between sales and profit, although a few high-sales orders still resulted in losses.

## 6.6 Box Plot

### Business Question

Are there any unusual (outlier) sales values?

```python
plt.figure(figsize=(8,5))

plt.boxplot(df["sales"])

plt.title("Box Plot of Sales")
plt.ylabel("Sales")

plt.show()
```

### Observation

The box plot reveals several high-value outliers in the sales data, confirming the presence of unusually large transactions.
